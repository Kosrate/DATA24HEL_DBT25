#### Group multiple operations in a transaction

❓ What is ACID?

✔️ Answer:
ACID stands for Atomicity, Consistency, Isolation, and Durability. These properties ensure reliable processing of database transactions:

- **Atomicity**: All operations within a transaction are completed successfully, or none are.
- **Consistency**: Transactions bring the database from one valid state to another.
- **Isolation**: Transactions are executed in isolation from one another.
- **Durability**: Once a transaction is committed, it remains so, even in the event of a system failure.

❓ Which of the three datasets (product catalog, photos and videos, business data) is transactional?

✔️ Answer: Business data is typically transactional as it often involves operations that require ACID properties to ensure data integrity and consistency.

## Integrate data with Azure Data Factory or Azure Synapse Pipeline
### Understand Azure Data Factory

❓ Which part of the symphony orchestra does Azure Data Factory play, according to the analogy?

✔️ Answer: Azure Data Factory plays the role of the conductor in the symphony orchestra, orchestrating and coordinating the movement of data between various data stores and processing services.

#### Describe data integration patterns

❓ What does ETL stand for?

✔️ Answer: ETL stands for Extract, Transform, Load.

❓ What does ELT stand for?

✔️ Answer: ELT stands for Extract, Load, Transform.

❓ What is the difference between ETL and ELT?

✔️ Answer: 
ETL (Extract, Transform, Load) and ELT (Extract, Load, Transform) are two different data integration patterns:

- **ETL**: Data is first extracted from source systems, then transformed into the desired format, and finally loaded into the target system. This approach is typically used in traditional data warehousing environments where transformations are performed before loading data into the data warehouse.

- **ELT**: Data is first extracted from source systems, then loaded into the target system, and finally transformed within the target system. This approach is commonly used in modern data processing environments, such as cloud-based data lakes and data warehouses, where the target system has the capability to perform transformations efficiently.

## Large-scale data processing with Azure Data Lake Storage Gen2
### Introduction to Azure Data Lake Storage Gen2
#### Understand Azure Data Lake Storage Gen2

❓ Which format is data stored in in a data lake?

✔️ Answer:
1. Structured data: CSV, Parquet, ORC
2. Semi-structured data: JSON, Avro
3. Unstructured data: Text files, images, videos

❓ Which factors should influence the planning of a data lake?

✔️ Answer:
1. Data Volume: The amount of data you expect to store and process.
2. Data Variety: The types of data (structured, semi-structured, unstructured) you will be handling.
3. Data Velocity: The speed at which data is ingested, processed, and analyzed.
4. Data Governance: Policies and procedures for data management, including data quality, security, and compliance.
5. Scalability: The ability to scale storage and processing resources as data volume grows.
6. Cost: Budget considerations for storage, processing, and data management.
7. Integration: Compatibility with existing data sources, tools, and workflows.
8. Performance: Requirements for data retrieval, processing speed, and query performance.
9. Security: Measures to protect data from unauthorized access and ensure data privacy.
10. Accessibility: Ease of access for data scientists, analysts, and other stakeholders.
